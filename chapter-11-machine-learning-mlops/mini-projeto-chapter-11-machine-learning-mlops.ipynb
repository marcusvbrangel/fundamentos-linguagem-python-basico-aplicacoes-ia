### 01 - Definição do Problema de Negõcio

É a fundação do projeto. Nesta etapa, traduzimos uma necessidade da empresa em um objetivo claro para a Ciência de Dados. Definimos o que queremos resolver (automatizar, classificação de reviews de produtos, por exemplo), porque é importante (reduzir custos, agilizar a tomada de decisão) e como o sucesso será medido.

**Definição:**

Uma empresa de e-commerce deseja automatizar a análise de sentimentos de seus clientes. Atualmente, a análise é feita manualmente, o que é um processo lento. caro e que não escala com o volume de reviews recebidos diariamente.

**Objetivo:**

Criar um modelo Machine Learning que classifique automaticamente os reviews de produtos como **"positivo"** ou **"negativo"**.

**Benefícios Esperados:**

- **Eficiência:** Reduzir o tempo e o custo da análise de feedback;
- **Tomada de Decisão Rápida**: Permitir que as equipes de produtos e marketing identifiquem rapidamente produtos com problemas ou oportunidades de melhoria;
- **Priorização**: Direcionar reviews negativos para a equipe de suporte ao cliente de forma prioritária, melhorando a experiância do consumidor.
