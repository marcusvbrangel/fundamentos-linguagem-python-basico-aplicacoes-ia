### 01 - Classes e Objetos

```python
# definicao da classe (modelo)

class Carro:

    def __init__(self, marca, modelo, ano):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.ligagdo = False

    def ligar(self):

        if not self.ligagdo:
            self.ligagdo = True
            print(f"O {self.modelo} foi ligado.")
        else:
            print(f"O {self.modelo} já estava ligado.")
        
    def desligar(self):

        if self.ligagdo:
            self.ligagdo = False
            print(f"O {self.modelo} foi desligado")
        else:
            print(f"O {self.modelo} já estava desligado.")

    def exibir_informacoes(self):
        print(f"Marca: {self.marca}, Modelo: {self.modelo}, Ano: {self.ano}")
```

```python
# criacao do objeto (instancia da classe Carro)
fusca = Carro("Volkswagen", "Fusca", 1985)
```

```python
# usando o objeto (fusca)
fusca.exibir_informacoes()

fusca.ligar()

fusca.ligar()

fusca.desligar()
```

### 02 - Encapsulamento

```python
# definicao da classe (modelo)

class Carro:

    def __init__(self, marca, modelo, ano):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.ligagdo = False
        self._velocidade = 0         # atributo protegido, nao deve ser acessado diretamente
        self.__potencia = 300        # atributo privado (name mangling), nao deve ser acessado diretamnte

    def ligar(self):

        if not self.ligagdo:
            self.ligagdo = True
            print(f"O {self.modelo} foi ligado.")
        else:
            print(f"O {self.modelo} já estava ligado.")
        
    def desligar(self):

        if self.ligagdo:
            self.ligagdo = False
            print(f"O {self.modelo} foi desligado")
        else:
            print(f"O {self.modelo} já estava desligado.")

    def exibir_informacoes(self):
        print(f"Marca: {self.marca}, Modelo: {self.modelo}, Ano: {self.ano}")

    def get_velocidade(self):
        return self._velocidade

    def acelerar(self, valor):
        if (valor > 0):
            self._velocidade += valor
            print(f"O {self.modelo} acelerou para {self._velocidade} Km/h")
        else:
            print("O valor da aceleração deve ser positivo")

    def frear(self, valor):
        if (valor > 0):
            self._velocidade -= valor
            if (self._velocidade < 0):
                self._velocidade = 0
            print(f"O {self.modelo} freou para {self._velocidade} km/h")
        else:
            print("O valor da frenagem deve ser positivo")            

    def exibir_potencia(self):
        return self.__potencia
```

```python
# criacao do objeto (insancia da classe Carro)
opala = Carro("Chevrolet", "Opala", 1980)
```

```python
# ATENÇÃO: 

# observe que o atributo __potencia (com dois underscor) nao pode ser acessado
# print(opala.__potencia)      # AttributeError: 'Carro' object has no attribute '__potencia'

# mas o atributo _velocidade (com somente um undescor) pode ser acessado
print(opala._velocidade)
```

```python
# acesso direto (nao recomendado, mas possivel)
opala._velocidade = 200 # isso quebra o encapsulamento!
print(f"Velocidade alterada diretamente: {opala._velocidade} km/h")
```

**ATENÇÃO**: A célula acima funciona porque, em Python, o uso de um único sublinhado no início de um atributo (como _velocidade) é apenas uma convenção de programação, não uma regra imposta pela linguagem. Ou seja, o sublinhado indica para outros desenvolvedores que aquele atributo é considerado "protegido" e não deve ser acessado diretamente fora da classe, mas o interpretador Python não impede que você o modifique. Python aceita normalmente, sobrescrevendo o valor interno do atributo, mesmo que isso quebre a lógica de encapsulamento. Diferente de linguagens como Java ou C++, onde o modificador private de fato impede o acesso externo, em Python a proteção é baseada em boas práticas e disciplina do programador.

Se a intenção fosse dificultar ainda mais o acesso direto, poderia ser usado duplo sublinhado (__horsepower), que aciona um processo chamado name mangling. Isso não torna o atributo verdadeiramente privado, mas altera o nome interno e dificulta acessá-lo acidentalmente, embora ainda seja possível com alguma insistência.

Resumindo

- _atributo → apenas convenção, acesso é permitido.

- __atributo → Python aplica name mangling, mudando o nome interno do atributo para _NomeDaClasse__atributo. Isso dificulta o acesso, mas ainda é possível se você souber o nome interno.

Ou seja, em Python o encapsulamento é mais cultural do que técnico.

```python
# usando o objeto opala
opala.acelerar(50)
print(f"Velocidade atual: {opala.get_velocidade()} km/h")
opala.frear(20)
print(f"Velocidade atual: {opala.get_velocidade()} km/h")
```

### 03 - Herança

```python
# super classe (classe pai)
class Veiculo:

    # metodo construtor da classe pai
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo
        self.ligado = False

    def ligar(self):
        self.ligado = True
        print(f"O {self.modelo} foi ligado")

    def desligar(self):
        self.ligado = False
        print(f"O {self.modelo} foi desligado")




# subclasse (classe filha)
class Carro(Veiculo):

    # metodo construtor da classe filha
    def __init__(self, marca, modelo, portas):
        # chama o construtor da classe pai
        super().__init__(marca, modelo)
        self.portas = portas

    def exibir_info_carro(self):
        print(f"Carro: {self.marca} {self.modelo}, Portas: {self.portas}")


# subclasse (outra classe filha)
class Moto(Veiculo):

    # metodo construtor da classe filha
    def __init__(self, marca, modelo, cilindradas):
        # chama o construtor da classe pai
        super().__init__(marca, modelo)
        self.cilindrada = cilindradas

    def empinar(self):
        print(f"A moto {self.modelo} está empinando! Cuidado!")
```

```python
vectra = Carro("Chevrolet", "Vectra", 4)
```

```python
vectra.exibir_info_carro()
vectra.ligar()
```

```python
cb500 = Moto("Honda", "CB 500", 500)
```

```python
cb500.ligar()
cb500.empinar()
```

### 04 - Polimorfismo

```python
# super classe (classe pai)
class Veiculo:

    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def exibir_detalhes(self):
        print(f"Veículo genérico: {self.marca} {self.modelo}")



# subclasse (classe filha)
class Carro(Veiculo):

    def __init__(self, marca, modelo, portas):
        super().__init__(marca, modelo)
        self.portas = portas

    # sobrescrevendo o metodo da classe pai
    def exibir_detalhes(self):
        print(f"Carro: {self.marca} {self.modelo}, Portas: {self.portas}")
        # super().exibir_detalhes()    # chamada de um metodo da classe pai



# subclasse (classe filha)
class Moto(Veiculo):

    def __init__(self, marca, modelo, cilindradas):
        super().__init__(marca, modelo)
        self.cilindradas = cilindradas

    # sobrescrevendo o metodo da classe pai
    def exibir_detalhes(self):
        print(f"Moto: {self.marca} {self.modelo}, Cilindradas: {self.cilindradas}")
```

```python
# definicao da lista de diferentes tipos
veiculos = [
    Carro("Volkswagen", "Fusca", 4),
    Moto("Honda", "CB 500", 500),
    Veiculo("Caloi", "Caloi Cross")    # usando a classe pai diretamente   
]
```

```python
# o mesmo metodo "exibir_detalhes()" se comporta de forma diferente para cada objeto
for v in veiculos:
    v.exibir_detalhes()   # polimorfismo em acao!!!   
```

### 05 - Métodos Especiais (embutidos)

```python
# definicao da classe (modelo)

class Livro:

    def __init__(self, codigo, titulo, autor):
        
        # pass => instrucao vazia usada quando a sintaxe exige um 
        # bloco de codigo, mas que ainda nao temos
        # pass

        self.codigo = codigo
        self.titulo = titulo
        self.autor = autor

    # equivalente ao metodo "toString()" do java
    def __str__(self):
        return f"{self.codigo}, {self.titulo}, {self.autor}"
    
    # equivalente ao "equals()" do java
    def __eq__(self, other):
        if not isinstance(other, Livro):
            return NotImplemented
        return self.codigo == other.codigo
    
    # equivalente ao "hashcode" do java
    def __hash__(self):
        return hash(self.codigo)
    
    # permite o resultado __str__ ser exibido em lista de objetos
    def __repr__(self):
        return f"Livro({self.codigo}, {self.titulo!r}, {self.autor!r})"
```

```python
# criacao dos objetos
livro1 = Livro(1, "Dom Casmurro", "Machado de Assis")
livro2 = Livro(1, "Dom Casmurro", "Machado de Assis")
livro3 = Livro(2, "Memorias Postumas", "Machado de Assis")
```

```python
# testa __str__
print(str(livro1))
print(livro1)  # print tambem usa __str__
```

```python
# testa __eq__
print(livro1 == livro2)  # True
print(livro1 == livro3)  # False
```

```python
# testa __hash__
print(hash(livro1))
print(hash(livro2))
print(hash(livro3))
```

```python
# testa uso em set
conjunto = {livro1, livro2, livro3}
print(conjunto)
print(len(conjunto))  # deve dar 2, porque livro1 e livro2 sao iguais
```

```python
try:

    numero = int(input("Digite um número:"))
    resultado = 10 / numero
    print("Resultado: ", resultado)

except ValueError:
    print("Erro: Você não digitou um número válido")

except ZeroDivisionError:
    print("Erro: não é possível dividir por zero")
```
